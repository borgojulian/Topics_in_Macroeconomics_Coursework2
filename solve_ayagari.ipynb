```python
import quantecon as qe
import matplotlib.pyplot as plt
import numpy as np
import jax
import jax.numpy as jnp
from typing import NamedTuple
from scipy.optimize import bisect
```

```python
@jax.jit
def compute_stationary(P):
    n = P.shape[0]
    I = jnp.identity(n)
    O = jnp.ones((n, n))
    A = I - jnp.transpose(P) + O
    return jnp.linalg.solve(A, jnp.ones(n))
```

```python
Pi_arg = [[0.7, 0.1, 0.1, 0.1], 
                 [0.1, 0.7, 0.1, 0.1],
                 [0.1, 0.1, 0.7, 0.1],
                 [0.1, 0.1, 0.1, 0.7]]

z_grid_arg = [1, 0.7, 0.5, 0.2]

z_grid, Pi = map(jnp.array, (z_grid_arg, Pi_arg))

```

```python
P = compute_stationary(Pi)
P
```
